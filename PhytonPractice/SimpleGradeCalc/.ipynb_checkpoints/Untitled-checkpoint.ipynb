```python
while True:
    midterm = float(input("Midterm?: "))
    final = float(input("Final?: "))

    average = (midterm * 0.4) + (final * 0.6)

    if average >= 85:
        result = "Passed with AA"
    elif average >= 70:
        result = "Passed with BB"
    elif average >= 50:
        result = "Passed with CC"
    else:
        result = "Failed"

    print(result)

    sentinel = input("Do you want to continue? (-1 to exit, any button to continue): ")
    if sentinel == "-1":
        break
```
